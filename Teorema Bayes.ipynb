# Ejemplo de red bayesiana

Simulación para determinar el rendimiento académico de un estudiante según sus horas de estudio y asistencia.

      HorasEstudio        Asistencia
             \             /
              \           /
               Rendimiento


```python
#!pip install pgmpy
```

```python
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination
```

```python
# crear la red bayesiana
model = DiscreteBayesianNetwork([
    ("HorasEstudio", "Rendimiento"),
    ("Asistencia", "Rendimiento")
])
```

```python
# crear la tabla de probabilidades para la variable HorasEstudio

# que puede 60% de probabilidad de que las horas de estudio sean altas
# y 40% de probabilidad de que las horas de estudio sean bajas

cpd_horas_estudio = TabularCPD(
    variable="HorasEstudio",
    variable_card=2,
    values=[[0.6], [0.4]],
    state_names={"HorasEstudio": ["Alta", "Baja"]}
)
```

```python
# crear la tabla de probabilidades para la variable Asistencia

# que puede 70% de probabilidad de que la asistencia sea alta
# y 30% de probabilidad de que la asistencia sea baja

cpd_asistencia = TabularCPD(
    variable="Asistencia",
    variable_card=2,
    values=[[0.7], [0.3]],
    state_names={"Asistencia": ["Alta", "Baja"]}
)
```

```python
# crear la tabla de probabilidades condicionales para la variable Rendimiento
# es una hipotesis de incertidumbre,
# el rendimiento depende de las horas de estudio y de la asistencia

cpd_rendimiento = TabularCPD(
    variable="Rendimiento",
    variable_card=2,
    values=[[0.9, 0.7, 0.6, 0.2], [0.1, 0.3, 0.4, 0.8]],
    evidence=["HorasEstudio", "Asistencia"],
    evidence_card=[2, 2],
    state_names={
        "Rendimiento": ["Alto", "Bajo"],
        "HorasEstudio": ["Alta", "Baja"],
        "Asistencia": ["Alta", "Baja"]
    }
)
```

```python
# agregar al modelo las tablas de probabilidades condicionales
model.add_cpds(cpd_horas_estudio, cpd_asistencia, cpd_rendimiento)
model.check_model()
```

```python
print("Modelo de red bayesiana creado y verificado correctamente.")
print(model.edges())
```

```python
# Inferencia en la red bayesiana
# motor o proceso de inferencia para al grafo
inferencia = VariableElimination(model)

resultado_rendimiento = inferencia.query(
    variables=["Rendimiento"],
    evidence={"HorasEstudio": "Alta", "Asistencia": "Alta"}
)

print("Probabilidad del rendimiento académico dado que las horas de estudio son altas y la asistencia es alta:")
print(resultado_rendimiento)
```
